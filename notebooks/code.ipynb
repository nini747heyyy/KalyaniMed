```python
import medmnist
from medmnist import INFO, PathMNIST
import matplotlib.pyplot as plt

# 1. Fetch metadata
info = INFO['pathmnist']
print("Classes:", info['label'])
print("Task:", info['task'])

# 2. Download and visualize a 5x5 grid of sample histology cells
dataset = PathMNIST(split="train", download=True)
montage_img = dataset.montage(length=5)

plt.figure(figsize=(8, 8))
plt.imshow(montage_img)
plt.axis("off")
plt.title("PathMNIST Histopathology Tissue Samples")
plt.show()
```
